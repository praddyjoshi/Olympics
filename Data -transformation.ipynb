```python
def func_cast(df, col_name):
    df_col=df.withColumn(col_name,F.flatten(F.transform(F.split(col_name, ','), lambda x: F.array(x))))
    return df_col

cast=func_cast(df_team,"athletes_codes")
cast2=func_cast(df_team,"coaches_codes")
display(cast)
```

```python
def to_snake(df):
    mapping = [F.col(i).alias(i.lower().replace(" ", "_")) for i in df.columns]
    return df.select(*mapping)

# Assuming df_athlet, df_coach, df_team, and df_medal are pandas DataFrames
# Convert them to Spark DataFrames before applying the to_snake function


df_athlet = to_snake(df_athlet)
df_coach = to_snake(df_coach)
df_team = to_snake(df_team)
df_medal = to_snake(df_medal)

display(df_athlet)
```
